```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# Loading the datasets

ecom = pd.read_csv(r"C:\Users\arwaa\Downloads\pakistani_ecommerce_dataset.csv")

student = pd.read_csv(r"C:\Users\arwaa\Downloads\student_lifestyle_dataset.csv")

print("Both datasets loaded successfully!")
```

```python
# Display first 5 rows of E-commerce dataset

display(ecom.head())
```

```python
# Display first 5 rows of Student dataset

display(student.head())
```

```python
# Check dataset information

print("E-Commerce Dataset:")
ecom.info()

print("\nStudent Lifestyle Dataset:")
student.info()
```

```python
# Count orders in each product category

category_counts = ecom['product_category'].value_counts()

print(category_counts)

# Find the most popular category

print("\nMost Popular Product Category:")
print(category_counts.idxmax())
```

```python
# Calculate average order value

average_order = ecom['order_value'].mean()

print("Average Order Value:", round(average_order, 2))
```

```python
# Count orders by city

city_orders = ecom['city'].value_counts()

print(city_orders)

# Find the city with the highest orders

print("\nCity with Highest Orders:")
print(city_orders.idxmax())
```

```python
# Count payment methods

payment_counts = ecom['payment_method'].value_counts()

print(payment_counts)

# Find the most used payment method

print("\nMost Used Payment Method:")
print(payment_counts.idxmax())
```

```python
# Calculate average order value by city

city_average = ecom.groupby('city')['order_value'].mean()

print(city_average)

# Create bar chart

plt.figure(figsize=(10, 6))

city_average.plot(kind='bar', color='skyblue')

plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value (PKR)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Calculate average study hours

average_study = student['study_hours'].mean()

print("Average Study Hours:", round(average_study, 2))
```

```python
# Calculate average sleep hours by gender

average_sleep = student.groupby('gender')['sleep_hours'].mean()

print(average_sleep)

# Find who sleeps more

print("\nGender with More Average Sleep:")
print(average_sleep.idxmax())
```

```python
# Calculate correlation

correlation = student['screen_time'].corr(
    student['sleep_hours']
)

print("Correlation:", round(correlation, 3))
```

```python
# Create scatter plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=student,
    x='screen_time',
    y='sleep_hours'
)

plt.title('Relationship Between Screen Time and Sleep Hours')
plt.xlabel('Screen Time')
plt.ylabel('Sleep Hours')

plt.tight_layout()
plt.show()
```

```python
# Draw histogram of daily transport cost

plt.figure(figsize=(8, 5))

plt.hist(
    student['daily_transport_cost'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.title('Distribution of Daily Transport Cost')
plt.xlabel('Daily Transport Cost (PKR)')
plt.ylabel('Number of Students')

plt.tight_layout()
plt.show()
```

```python
# Calculate average internet usage by gender

internet_gender = student.groupby('gender')['internet_usage'].mean()

print(internet_gender)

# Draw bar chart

plt.figure(figsize=(8, 5))

internet_gender.plot(
    kind='bar',
    color=['pink', 'skyblue']
)

plt.title('Average Internet Usage by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Internet Usage')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
